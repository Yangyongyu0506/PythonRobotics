```python
import numpy as np

phi_vec = np.array([0.3, 0.4, 0.5])

def hat(vec: np.ndarray) -> np.ndarray:
    """Compute the skew-symmetric matrix (hat operator) of a 3D vector."""
    return np.array([[0, -vec[2], vec[1]],
                     [vec[2], 0, -vec[0]],
                     [-vec[1], vec[0], 0]])

def vee(mat: np.ndarray) -> np.ndarray:
    """Compute the vector (vee operator) from a skew-symmetric matrix."""
    return np.array([mat[2, 1], mat[0, 2], mat[1, 0]])


def rodrigues(mat: np.ndarray) -> np.ndarray:
    """Computer the rotation matrix from a skew-symmetric matrix using Rodrigues' formula."""
    theta = np.linalg.norm(vee(mat))
    if theta < 1e-6:
        return np.eye(3)
    else:
        return np.eye(3) + (np.sin(theta) / theta) * mat + ((1 - np.cos(theta)) / (theta ** 2)) * (mat @ mat)

R1 = rodrigues(hat(phi_vec))
print(R1)
```

```python
Rz = rodrigues(hat(np.array([0, 0, phi_vec[2]])))
Ry = rodrigues(hat(np.array([0, phi_vec[1], 0])))
Rx = rodrigues(hat(np.array([phi_vec[0], 0, 0])))
R2 = Rz @ Ry @ Rx
print(R2)
```

```python
print(R1 @ R2.T)  # Should be close to identity matrix
```

```python
phi_vec_mini = phi_vec * 0.01
R1_mini = rodrigues(hat(phi_vec_mini))
R2_mini = rodrigues(hat(np.array([0, 0, phi_vec_mini[2]]))) @ \
            rodrigues(hat(np.array([0, phi_vec_mini[1], 0]))) @ \
            rodrigues(hat(np.array([phi_vec_mini[0], 0, 0])))
print(R1_mini @ R2_mini.T)
```
